## 1. Conectar ao banco sqlite em memória

```python
import sqlite3

conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

print("Conectado ao banco sqlite em memoria")
```

```python
cursor.execute("""DROP TABLE IF EXISTS [tcc]""")
cursor.execute("""DROP TABLE IF EXISTS [aluno]""")
cursor.execute("""DROP TABLE IF EXISTS [professor]""")
```

## 2. Criar uma tabela de exemplo

```python
cursor.execute("""
CREATE TABLE aluno (
    id INTEGER PRIMARY KEY,
    matricula VARCHAR(250) UNIQUE NOT NULL,
    nome VARCHAR(250) NOT NULL,
    email VARCHAR(250) NOT NULL,
    telefone VARCHAR(250) NOT NULL);
""")

print("Tabela aluno criada com sucesso.")
```

```python
cursor.execute("""
CREATE TABLE professor (
    id INT PRIMARY KEY,
    nome VARCHAR(255) NOT NULL,
    email VARCHAR(250) NOT NULL,
    orientador BOOLEAN NOT NULL,
    titulacao VARCHAR(250) NOT NULL,
    curso VARCHAR(250),
    instituicao VARCHAR(250),
    externo BOOLEAN NOT NULL);
""")

print("Tabela professor criada com sucesso.")
```

```python
cursor.execute("""
CREATE TABLE tcc (
    id INT PRIMARY KEY,
    id_orientador INT,
    id_aluno INT,
    titulo VARCHAR(250) NOT NULL,
    FOREIGN KEY (id_orientador) REFERENCES professor(id)
    ON DELETE NO ACTION
    ON UPDATE CASCADE,
    FOREIGN KEY (id_aluno) REFERENCES aluno(id)
    ON DELETE NO ACTION
    ON UPDATE CASCADE);
""")

print("Tabela tcc criada com sucesso.")
```

# Alterando uma tabela

```python
### Antes da alteração, vamos verificar o esquema da tabela aluno
```

```python
print("Esquema da tabela aluno ANTES do ALTER TABLE")

cursor.execute("PRAGMA table_info(aluno)")
for column in cursor.fetchall():
    print(column)
```

### Adicionar uma nova coluna 'idade' à tabela aluno

```python
cursor.execute("ALTER TABLE aluno ADD COLUMN idade INT;")
conn.commit()

print("Coluna 'idade' adicionada à tabela aluno com sucesso.")
```

### Após a alteração, vamos verificar o esquema novamente

```python
print("Esquema da tabela aluno DEPOIS do ALTER TABLE: ")
cursor.execute("PRAGMA table_info(aluno)")
for column in cursor.fetchall():
    print(column)
```

### E também ver os dados para confirmar que a nova coluna existe (com valores NULL inicialmente)

```python
print("Dados da tabela aluno com a nova coluna: ")
cursor.execute("SELECT * FROM aluno")
for aluno in cursor.fetchall():
    print(aluno)
```

## Inserindo um registro

## Inserindo múltiplos registros

```python
# Preparar os dados para inserção em massa
dados_novos_alunos = [
    ('20260033', 'Bruno Costa', 'bruno@aluno.osorio.ifrs.edu.br', '(51)991112222'),
    ('20260034', 'Carla Dias', 'carla@aluno.osorio.ifrs.edu.br', '(51)993334444'),
    ('20260035', 'Daniela Souza', 'daniela@aluno.osorio.ifrs.edu.br', '(51)995556666')
]

# A query de INSERT é a mesma, mas agora ela será executada para cada item da lista
cursor.executemany("""
INSERT INTO aluno (matricula, nome, email, telefone) VALUES (?, ?, ?, ?)
""", dados_novos_alunos)

conn.commit()
print(f"{cursor.rowcount} novos alunos inseridos com sucesso!")
```

```python
dados_novos_professores = [
    (1, 'Carlos', 'carlos.p@ifrs.edu.br', True, 'Doutorado', 'Engenharia', 'IFRS', False),
    (2, 'Maria', 'maria.s@ifrs.edu.br', True, 'Mestrado', 'Informática', 'IFRS', False),
    (3, 'João', 'joao.e@outra.org', False, 'Doutorado', 'Informática', 'Outra Instituição', True)
]

cursor.executemany("""
INSERT INTO professor (id, nome, email, orientador, titulacao, curso, instituicao, externo)
VALUES (?, ?, ?, ?, ?, ?, ?, ?)
""", dados_novos_professores)

conn.commit()
print(f"{cursor.rowcount} novos professores inseridos com sucesso!")
```

```python
dados_novos_tccs = [
    (1, 1, 'Estudo de Caso: Aplicação de IA na Agricultura'),
    (2, 2, 'Desenvolvimento de um Sistema de Gerenciamento de Projetos Web'),
    (3, 1, 'Análise Comparativa de Frameworks Frontend JavaScript')
]

# Assumindo que o ID do TCC será gerado automaticamente (INTEGER PRIMARY KEY)
cursor.executemany("""
INSERT INTO tcc (id_orientador, id_aluno, titulo)
VALUES (?, ?, ?)
""", dados_novos_tccs)

conn.commit()
print(f"{cursor.rowcount} novos TCCs inseridos com sucesso!")
```

```python
cursor.execute("""
SELECT * FROM aluno;
""")

linhas = cursor.fetchall()
for linha in linhas:
    print(linha)
```

```python
cursor.execute("""
SELECT
    A.nome AS nome_aluno,
    P.nome AS nome_orientador,
    T.titulo AS titulo_tcc
FROM
    tcc AS T
JOIN
    aluno AS A ON T.id_aluno = A.id
JOIN
    professor AS P ON T.id_orientador = P.id
WHERE
    T.titulo = 'Análise Comparativa de Frameworks Frontend JavaScript'
""")

resultado = cursor.fetchone() # Ou fetchall() se espera múltiplos resultados

if resultado:
    print(f"Aluno: {resultado[0]}, Orientador: {resultado[1]}, TCC: {resultado[2]}")
else:
    print("TCC não encontrado.")
```

## Alterando registros

```python
cursor.execute("""UPDATE aluno SET idade = 0;""")
conn.commit()
print("Coluna idade atualizada para 0 em todos os registros")

#Verificar os dados atualizados
print("Dados da tabela aluno APÓS a atualização da coluna idade")
cursor.execute("SELECT * FROM aluno")
for aluno in cursor.fetchall():
    print(aluno)
```

```python
cursor.execute("UPDATE professor SET titulacao = 'Doutorado' WHERE nome = 'Maria';")
conn.commit()
print("Titulação alterada com sucesso.")

#Verificar o registro atualizado
print("Dados da professora Maria APÓS a atualização.")
cursor.execute("SELECT * FROM professor WHERE nome = 'Maria';")
for professor in cursor.fetchall():
    print(professor)
```
