```python
import psycopg2

# Conectando agora na porta 5433
connection = psycopg2.connect(
    user="postgres",
    password="senha123",
    host="127.0.0.1",
    port="5433", # <--- Mudança crucial aqui
    database="db_series"
)

cursor = connection.cursor()

# Verificando a versao do db
cursor.execute("SELECT version();")
db_version = cursor.fetchone()
print("Conectado com sucesso!")
print(f"Versão: {db_version}")

# Criando a tabela
cursor.execute("""
    CREATE TABLE IF NOT EXISTS alunos (
        id SERIAL PRIMARY KEY,
        nome VARCHAR(100) NOT NULL
    );
""")

# Inserindo o dado
cursor.execute("INSERT INTO alunos (nome) VALUES ('Joao');")
connection.commit()
cursor.execute("INSERT INTO alunos (nome) VALUES ('Maria');")
connection.commit()

# Buscando os dados
cursor.execute("SELECT * FROM alunos;")
resultado = cursor.fetchall()

print("Dados recuperados:")
for linha in resultado:
    print(linha)

cursor.close()
connection.close()
```
